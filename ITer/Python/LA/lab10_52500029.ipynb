```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# cau 9
def draw_house(pts, transformed_pts, title):
    plt.plot(pts[0], pts[1], 'b--', label='Original')
    plt.plot(transformed_pts[0], transformed_pts[1], 'r-', label='Transformed')
    plt.legend()
    plt.title(title)
    plt.grid()
    plt.axis('equal')
    plt.show()

house = np.array([
    [0, 2, 2, 1, 0, 0],
    [0, 0, 2, 3, 2, 0]
])

house_a = house + np.array([[2], [4]])
draw_house(house, house_a, '9a. Translation tx=2, ty=4')

alpha = np.pi/3
R = np.array([[np.cos(alpha), -np.sin(alpha)], [np.sin(alpha), np.cos(alpha)]])
house_b = R @ house
draw_house(house, house_b, '9b. Rotation alpha=pi/3')

S = np.array([[2, 0], [0, 3]])
house_c = S @ house
draw_house(house, house_c, '9c. Scaling Sx=2, Sy=3')

Sh_x = np.array([[1, 0.5], [0, 1]])
house_dx = Sh_x @ house
draw_house(house, house_dx, '9d. Shear along x (0.5)')

Sh_y = np.array([[1, 0], [-1.5, 1]])
house_dy = Sh_y @ house
draw_house(house, house_dy, '9d. Shear along y (-1.5)')
```

```python
# cau 10
P = [1, 1]
Q = [3, 1]
R = [1, 3]
A = np.array([P, Q, R, P]).T

I = np.eye(2)
T = -I
A_transformed = T @ A

plt.plot(A[0], A[1], 'b--', label='Original PQR')
plt.plot(A_transformed[0], A_transformed[1], 'r-', label='Image under -I')
plt.legend()
plt.title('Exercise 10')
plt.axis('equal')
plt.show()
```

```python
# cau 11
F_pts = np.array([
    [0, 0, 3, 3, 1, 1, 2, 2, 1, 1, 0],
    [0, 5, 5, 4, 4, 3, 3, 2, 2, 0, 0]
])
w = np.ones((1, F_pts.shape[1]))
F_3d = np.vstack([F_pts, w])

T1 = np.array([
    [-1, 0, 0],
    [0, -1, 0],
    [0, 0, 1]
])
T2 = np.array([
    [0.25, 0, 0],
    [0, 0.25, 0],
    [0, 0, 1]
])

T3 = np.array([[1, 0, 5/4], [0, 1, -5/4], [0, 0, 1]])
T4 = np.array([[1/4, 0, 5/4], [0, 1/4, -5/4], [0, 0, 1]])
T5 = np.array([[-1/4, 0, 0], [0, -1/4, 0], [0, 0, 1]])
T6 = np.array([[-1/4, 0, -5/4], [0, -1/4, 5/4], [0, 0, 1]])
T7 = np.array([[1, -1, 5/4], [0, 0, 5/4], [0, 0, 1]])

def draw_T(T, F, F_orig, title):
    F_new = T @ F_orig
    plt.plot(F_orig[0], F_orig[1], 'b--', label='Original')
    plt.plot(F_new[0], F_new[1], 'r-', label='Transformed')
    plt.legend()
    plt.title(title)
    plt.axis('equal')
    plt.show()

draw_T(T1, F_pts, F_3d, '11. T1')
draw_T(T2, F_pts, F_3d, '11. T2')
draw_T(T3, F_pts, F_3d, '11. T3')
draw_T(T4, F_pts, F_3d, '11. T4')
draw_T(T5, F_pts, F_3d, '11. T5')
draw_T(T6, F_pts, F_3d, '11. T6')
draw_T(T7, F_pts, F_3d, '11. T7')
```

```python
# cau 12
x12 = np.array([2015, 2016, 2017, 2018, 2019])
y12 = np.array([12, 19, 29, 37, 45])

A12 = np.vstack([x12, np.ones(len(x12))]).T
w12 = np.linalg.lstsq(A12, y12, rcond=None)[0]
print(f"Equation: y = {w12[0]:.5f}x + {w12[1]:.5f}")

sales_2020 = w12[0] * 2020 + w12[1]
print(f"Estimated sales in 2020: {sales_2020:.5f}")

plt.scatter(x12, y12, c='r')
plt.plot(x12, w12[0]*x12 + w12[1], c='b')
plt.scatter(2020, sales_2020, c='g', marker='x', s=100, label='2020 Estimate')
plt.legend()
plt.title('Exercise 12')
plt.show()
```
