```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import math

x, y, t = sp.symbols('x y t')
```

```python
# Cau 1
def ex1():
    # a
    f_a = x**3 + 2*x**2 + 3
    res_a = sp.integrate(f_a, (x, 1, 2))
    print("Cau 1a:", res_a)
    
    # b
    f_b = 1/x**3 + 1/x**2 + x*sp.sqrt(x)
    res_b = sp.integrate(f_b, (x, 1, 4))
    print("Cau 1b:", res_b)

    # c
    f_c = (x**3 + x*sp.sqrt(x) + x)/x**2
    res_c = sp.integrate(f_c, (x, 1, 4))
    print("Cau 1c:", res_c)

    # d
    f_d = 2/x + x**3
    res_d = sp.integrate(f_d, (x, 1, 2))
    print("Cau 1d:", res_d)

    # e
    f_e = x**2*(1/x + 2*x)
    res_e = sp.integrate(f_e, (x, 1, 2))
    print("Cau 1e:", res_e)

    # f
    f_f = (sp.sqrt(x) - 1)*(x + sp.sqrt(x) + 1)
    res_f = sp.integrate(f_f, (x, 0, 1))
    print("Cau 1f:", res_f)

    # g
    f_g = 1 - 2/sp.sin(x)**2
    res_g = sp.integrate(f_g, (x, sp.pi/4, sp.pi/2))
    print("Cau 1g:", res_g)

    # h
    f_h = 1/(sp.sin(x)**2 * sp.cos(x)**2)
    res_h = sp.integrate(f_h, (x, sp.pi/6, sp.pi/4))
    print("Cau 1h:", res_h)

    # i
    f_i = sp.exp(x)*(1 - sp.exp(-x)/sp.cos(x)**2)
    res_i = sp.integrate(f_i, (x, 0, sp.pi/4))
    print("Cau 1i:", res_i)

    # j
    f_j = sp.exp(x)*(2 + sp.exp(-x)/sp.exp(x))
    res_j = sp.integrate(f_j, (x, 0, sp.log(2)))
    print("Cau 1j:", res_j)

    # k
    f_k = 2**x + 2/x
    res_k = sp.integrate(f_k, (x, 1, 2))
    print("Cau 1k:", res_k)

    # l
    f_l = x**2*(x - 1)**2
    res_l = sp.integrate(f_l, (x, 0, 1))
    print("Cau 1l:", res_l)

    # m
    f_m = 1/(x*(x + 1))
    res_m = sp.integrate(f_m, (x, 1, 2))
    print("Cau 1m:", res_m)

    # n
    f_n = sp.Abs(1 - x)
    res_n = sp.integrate(f_n, (x, 0, 2))
    print("Cau 1n:", res_n)

    # o
    res_o = sp.integrate(2*x - x**2, (x, 0, 2)) + sp.integrate(x**2 - 2*x, (x, 2, 3))
    print("Cau 1o:", res_o)

    # p
    f_p = sp.sqrt(x**2 - 3*x + 2)
    res_p = sp.integrate(f_p, (x, 2, 4))
    print("Cau 1p:", res_p)

    # q
    res_q = sp.sqrt(2) * sp.integrate(sp.cos(x), (x, 0, sp.pi/2)) - sp.sqrt(2) * sp.integrate(sp.cos(x), (x, sp.pi/2, sp.pi))
    print("Cau 1q:", res_q)

ex1()
```

```python
# Cau 2
def plot_func(f, interval, title):
    f_num = sp.lambdify(x, f, 'numpy')
    x_vals = np.linspace(float(interval[0]), float(interval[1]), 1000)
    y_vals = f_num(x_vals)
    plt.plot(x_vals, y_vals)
    plt.title(title)
    plt.fill_between(x_vals, y_vals, alpha=0.2)
    plt.show()

def ex2():
    # a
    f_a = x**3 - 3*sp.sin(x)*sp.cos(x)
    res_a = sp.integrate(f_a, (x, 0, sp.pi/2))
    print("Cau 2a:", res_a)
    plot_func(f_a, (0, np.pi/2), "Cau 2a")

    # b
    f_b = sp.sin(x**2)**2
    res_b = sp.integrate(f_b, (x, 0, 1))
    print("Cau 2b:", res_b.evalf())
    plot_func(f_b, (0, 1), "Cau 2b")

    # c
    g = x + 1
    f_c = sp.sqrt(1 + g.subs(x, x**2) + g**2)
    res_c = sp.integrate(f_c, (x, 0, 3))
    print("Cau 2c:", res_c.evalf())
    plot_func(f_c, (0, 3), "Cau 2c")

    # d
    f_d = x**2 * y
    res_d = sp.integrate(f_d, (x, 0, 3), (y, 1, 2))
    print("Cau 2d:", res_d)

    x_vals = np.linspace(0, 3, 100)
    y_vals = np.linspace(1, 2, 100)
    X, Y = np.meshgrid(x_vals, y_vals)
    Z = X**2 * Y
    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')
    ax.plot_surface(X, Y, Z, cmap='viridis')
    plt.title("Cau 2d")
    plt.show()

ex2()
```

```python
# Cau 3
def ex3():
    def solve_and_plot(f, a, b, name):
        avg = (1/(b - a)) * sp.integrate(f, (x, a, b))
        print(f"Cau 3{name} average value:", avg.evalf())
        
        f_num = sp.lambdify(x, f, 'numpy')
        x_vals = np.linspace(float(a), float(b), 100)
        y_vals = f_num(x_vals)
        plt.plot(x_vals, y_vals, label="f(x)")
        plt.axhline(float(avg.evalf()), color='r', linestyle='--', label="Average")
        plt.title(f"Cau 3{name}")
        plt.legend()
        plt.show()

    solve_and_plot(x**2 - 1, 0, sp.sqrt(3), 'a')
    solve_and_plot(-x**2/2, 0, 3, 'b')
    solve_and_plot(-3*x**2 - 1, 0, 1, 'c')
    solve_and_plot(x**2 - x, -2, 1, 'd')

ex3()
```

```python
# Cau 4
def ex4():
    # a
    f_a = x**2 * sp.cos(x)
    res_a = sp.integrate(f_a, (x, -4, 9))
    print("Cau 4a Area:", res_a.evalf())
    x_vals = np.linspace(-4, 9, 100)
    f_num = sp.lambdify(x, f_a, 'numpy')
    plt.plot(x_vals, f_num(x_vals))
    plt.fill_between(x_vals, f_num(x_vals), alpha=0.2)
    plt.title("Cau 4a")
    plt.show()

    # b
    a = 1/2
    f_b = sp.exp(-a * x**2)
    res_b = sp.integrate(f_b, (x, -sp.oo, sp.oo))
    print("Cau 4b Area:", res_b)
    x_vals = np.linspace(-5, 5, 100)
    f_num = sp.lambdify(x, f_b, 'numpy')
    plt.plot(x_vals, f_num(x_vals))
    plt.fill_between(x_vals, f_num(x_vals), alpha=0.2)
    plt.title("Cau 4b")
    plt.show()

ex4()
```

```python
# Cau 5
def ex5():
    v = 160 - 32*t
    displacement = sp.integrate(v, (t, 0, 8))
    print("Cau 5 Displacement:", displacement, "ft")

ex5()
```

```python

# Cau 6
def ex6():
    dc_dx = 1 / (2*sp.sqrt(x))
    cost = sp.integrate(dc_dx, (x, 1, 100))
    print("Cau 6 Cost:", cost, "dollars")

ex6()
```

```python
# Cau 7
def ex7():
    H = sp.sqrt(t + 1) + 5*t**(sp.Rational(1, 3))
    
    # a
    h_0 = H.subs(t, 0)
    h_4 = H.subs(t, 4)
    h_8 = H.subs(t, 8)
    print("Cau 7a:")
    print("Height at t=0:", h_0.evalf(), "ft")
    print("Height at t=4:", h_4.evalf(), "ft")
    print("Height at t=8:", h_8.evalf(), "ft")

    # b
    avg_h = (1/(8 - 0)) * sp.integrate(H, (t, 0, 8))
    print("Cau 7b Average height:", avg_h.evalf(), "ft")

ex7()
```

```python
# Cau 8
def ex8():
    def evaluate_program(f, a, b, name, integral_val):
        print(f"\nCau 8{name}")
        f_num = sp.lambdify(x, f, 'numpy')
        x_vals = np.linspace(float(a), float(b), 1000)
        y_vals = f_num(x_vals)
        plt.plot(x_vals, y_vals)
        plt.title(f"Cau 8{name}: f(x) on [{a}, {b}]")
        plt.show()
        for n in [4, 100, 200, 1000]:
            width = (b - a) / n
            midpoints = [a + width/2 + i*width for i in range(n)]
            f_values = [f_num(float(m)) for m in midpoints]
            avg_val = sum(f_values) / n
            print(f"n = {n:4d} | Average value of f(x) = {avg_val:.6f}")
        print(f"Expected Integral = {integral_val}, Expected Average = {float(integral_val / (b - a)):.6f}")
    evaluate_program(1 - x, 0, 1, 'a', 1/2)
    evaluate_program(x**2 + 1, 0, 1, 'b', 4/3)
    evaluate_program(sp.cos(x), -sp.pi, sp.pi, 'c', 0)
    evaluate_program(sp.Abs(x), -1, 1, 'd', 1)

ex8()
```
