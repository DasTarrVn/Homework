## Phần I. Lý Thuyết

### 1. Xác suất thực nghiệm

Khi lặp lại một phép thử độc lập $n$ lần và biến cố $A$ xuất hiện $k$ lần, xác suất thực nghiệm của $A$ là

$$\widehat{P}(A)=\frac{k}{n}. $$

Với phép lấy mẫu không hoàn lại, mỗi lần thí nghiệm phải bắt đầu lại từ đầy đủ các phần tử ban đầu. Dùng bộ sinh số ngẫu nhiên với `seed` giúp kết quả mô phỏng có thể tái lập.

```python
import itertools
import math
import random

def experimental_probability(event_count, number_of_trials):
    """Tính tần suất tương đối của một biến cố."""
    if number_of_trials <= 0:
        raise ValueError("Số lần thí nghiệm phải là số nguyên dương.")
    return event_count / number_of_trials
```

### 2. Tổ hợp và xác suất trên không gian mẫu hữu hạn

Số cách chọn $r$ phần tử từ $n$ phần tử mà không xét thứ tự là

$$\binom{n}{r}=\frac{n!}{r!(n-r)!}. $$

Nếu các kết quả sơ cấp đồng khả năng thì

$$P(A)=\frac{|A|}{|\Omega|}. $$

Trong Python, `itertools.combinations(S, r)` liệt kê các tổ hợp và `math.comb(n, r)` tính $\binom{n}{r}$.
