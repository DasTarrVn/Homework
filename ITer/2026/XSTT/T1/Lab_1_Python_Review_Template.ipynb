### 1. Kiểu dữ liệu cơ bản

#### 1.1. Số

Python phân biệt số nguyên (`int`) và số thực (`float`). Các toán tử thường dùng:

| Toán tử | Ý nghĩa | Ví dụ |
|---|---|---|
| `+`, `-`, `*` | cộng, trừ, nhân | `3 * 2` |
| `/` | chia thực | `7 / 2 == 3.5` |
| `//` | chia lấy phần nguyên | `7 // 2 == 3` |
| `%` | chia lấy dư | `7 % 2 == 1` |
| `**` | lũy thừa | `3 ** 2 == 9` |

Python không có toán tử `x++` hoặc `x--`; hãy dùng `x += 1` hoặc `x -= 1`.

```python
x = 3
print(type(x))
print(x + 1, x - 1, x * 2, x ** 2)

x += 1
x *= 2
print("x sau khi cập nhật:", x)

y = 2.5
print(type(y))
print(y, y + 1, y * 2, y ** 2)
```

### 2. Containers

#### 2.1. List

`list` là dãy có thứ tự, thay đổi được và có thể chứa các phần tử khác kiểu. Chỉ số bắt đầu từ `0`; chỉ số âm đếm từ cuối danh sách.

```python
xs = [3, 1, 2]
print(xs, xs[2], xs[-1])

xs[2] = "foo"
xs.append("bar")
print("Sau khi sửa và append:", xs)

last_item = xs.pop()
print("Phần tử lấy ra:", last_item)
print("Danh sách còn lại:", xs)
```

##### Slicing

Cú pháp tổng quát là `sequence[start:stop:step]`; vị trí `stop` không được lấy.

```python
nums = list(range(5))
print("nums      =", nums)
print("nums[2:4] =", nums[2:4])
print("nums[2:]  =", nums[2:])
print("nums[:2]  =", nums[:2])
print("nums[:-1] =", nums[:-1])
print("nums[::-1]=", nums[::-1])

nums[2:4] = [8, 9]
print("Sau phép gán slice:", nums)
```

##### Vòng lặp, `enumerate` và list comprehension

- `enumerate` dùng để duyệt qua một danh sách đồng thời lấy được cả index và value của từng phần tử.

```python
animals = ["cat", "dog", "monkey"]

for index, animal in enumerate(animals, start=1):
    print(f"#{index}: {animal}")

nums = list(range(5))
squares = [x**2 for x in nums]
even_squares = [x**2 for x in nums if x % 2 == 0]
print("Bình phương:", squares)
print("Bình phương số chẵn:", even_squares)
```

#### 2.2. Dictionary

`dict` lưu các cặp `key: value`.

```python
d = {"cat": "cute", "dog": "furry"}
print(d["cat"])
print("cat" in d)

d["fish"] = "wet"
print(d.get("monkey", "N/A"))

for animal, description in d.items():
    print(f"{animal}: {description}")

del d["fish"]
print(d)

even_number_to_square = {x: x**2 for x in range(5) if x % 2 == 0}
print(even_number_to_square)
```

#### 2.3. Set

`set` là tập hợp các phần tử **không trùng lặp**.

```python
animals = {"cat", "dog"}
animals.add("fish")
animals.add("cat")  # Không tạo phần tử trùng
print("Số phần tử:", len(animals))
print("fish" in animals)

animals.remove("cat")
print("Sau khi xóa:", animals)

from math import sqrt

integer_roots = {int(sqrt(x)) for x in range(30)}
print("Set comprehension:", integer_roots)
```

#### 2.4. Tuple

tuple là kiểu dữ liệu dùng để lưu nhiều giá trị trong một biến, tương tự list, nhưng không thể thay đổi sau khi tạo.

```python
coordinate = (5, 6)
lookup = {(x, x + 1): x for x in range(10)}

print(type(coordinate))
print(lookup[coordinate])
print(lookup[(1, 2)])
```

### 3. Hàm

Hàm được định nghĩa bằng `def`. Dùng `return` để trả kết quả; tham số có thể có giá trị mặc định.

```python
def sign(x):
    if x > 0:
        return "positive"
    if x < 0:
        return "negative"
    return "zero"


for number in [-1, 0, 1]:
    print(number, "->", sign(number))
```

### 4. Lớp và đối tượng

Một lớp gom dữ liệu và hành vi liên quan. `__init__` khởi tạo đối tượng; `self` tham chiếu đến chính đối tượng đang gọi phương thức.

```python
class Greeter:
    def __init__(self, name):
        self.name = name

    def greet(self, loud=False):
        message = f"Hello, {self.name}"
        print(message.upper() + "!" if loud else message)


greeter = Greeter("Fred")
greeter.greet()
greeter.greet(loud=True)
```
