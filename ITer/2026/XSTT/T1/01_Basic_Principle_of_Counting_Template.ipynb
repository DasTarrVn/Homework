### 1. Permutation

A **permutation** of a set is an ordered arrangement of its members. Let $S$ be a finite set containing $n$ objects, and select $k\le n$ objects.

- **Permutations with repetition:** each element of $S$ may occur repeatedly, up to $k$ times. The number of arrangements is $n^k$.
- **Permutations without repetition:** each selected element occurs only once. The number is

$$P_n^k=n(n-1)(n-2)\cdots(n-k+1)=\frac{n!}{(n-k)!}.$$

For example, let $E=\{a,b,c,d\}$. The ordered arrangements of three elements are

$(a,b,c)$, $(a,c,b)$, $(b,a,c)$, $(b,c,a)$, $(c,a,b)$, $(c,b,a)$,  
$(a,b,d)$, $(a,d,b)$, $(b,a,d)$, $(b,d,a)$, $(d,a,b)$, $(d,b,a)$,  
$(a,c,d)$, $(a,d,c)$, $(c,a,d)$, $(c,d,a)$, $(d,a,c)$, $(d,c,a)$,  
$(b,c,d)$, $(b,d,c)$, $(c,b,d)$, $(c,d,b)$, $(d,b,c)$, $(d,c,b)$.

Thus, $P_4^3=4\cdot3\cdot2=\frac{4!}{(4-3)!}=24$.

```python
import itertools

E = ('a', 'b', 'c', 'd')
k = 3
n = len(E)
permute_k = list(itertools.permutations(E, k))

print(f"{k}-permutations of {E}:")
for arrangement in permute_k:
    print(arrangement)
print(f"Size = {n}!/({n}-{k})! = {len(permute_k)}")
```

### 2. Combination

A **combination** is a selection of items for which the order of selection does not matter. A $k$-combination of a set $S$ is a subset containing $k$ distinct elements. Its number is

$$C_n^k=\binom{n}{k}=\frac{n(n-1)\cdots(n-k+1)}{k(k-1)\cdots1}=\frac{n!}{k!(n-k)!}.$$

This formula applies when $k\le n$; the count is zero when $k>n$.

For $E=\{a,b,c,d\}$, the groups of three are $(a,b,c)$, $(a,b,d)$, $(a,c,d)$, and $(b,c,d)$. Therefore,

$$C_4^3=\frac{4!}{3!(4-3)!}=4.$$

```python
E = ('a', 'b', 'c', 'd')
k = 3
n = len(E)
choose_k = list(itertools.combinations(E, k))

print(f"{k}-combinations of {E}:")
for group in choose_k:
    print(group)
print(f"Number of combinations = {n}!/({k}!({n}-{k})!) = {len(choose_k)}")
```

### 3. Urn problems

An urn contains 8 white balls, 6 black balls, and 9 red balls. Six balls are drawn without replacement.

1. How many ways can 6 balls be selected?
2. Enumerate all possible selections of 6 balls.
3. Enumerate the selections whose first and last stored balls are red.

Represent the white balls by `W1`, `W2`, ..., `W8`; the black balls by `B1`, ..., `B6`; and the red balls by `R1`, ..., `R9`. The sample space is the set of 6-combinations of the 23 distinct balls. Hence,

$$|U6|=C_{23}^{6}=\frac{23\cdot22\cdot21\cdot20\cdot19\cdot18}{6!}=100947.$$

```python
def cross(A, B):
    """Concatenate every item from collection A with every item from B."""
    return tuple(a + b for a in A for b in B)

urn = cross('W', '12345678') + cross('B', '123456') + cross('R', '123456789')
print(urn)
```

```python
# import itertools

U6 = list(itertools.combinations(urn, 6))
print(len(U6))
```

```python
def print_all_U6():
    for selection in U6:
        print(selection)

print_all_U6()
```

```python
U6_red_ends = [
    selection for selection in U6
    if selection[0][0] == 'R' and selection[-1][0] == 'R'
]
print(f"Selections whose first and last stored balls are red: {len(U6_red_ends)}")
```
