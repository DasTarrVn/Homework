### 1. DataFrame và cách tạo DataFrame mẫu

`DataFrame` là cấu trúc dữ liệu hai chiều do thư viện `pandas` cung cấp, tương tự một bảng gồm các dòng và cột.

Mỗi dòng được xác định bằng chỉ số nguyên (bắt đầu từ 0) hoặc bằng nhãn được đặt khi tạo DataFrame. Mỗi cột có một nhãn riêng; các cột có thể chứa những kiểu dữ liệu khác nhau.

Tạo DataFrame từ một dictionary gồm hai cột `numbers` và `colors`. Mỗi khóa là tên cột, còn giá trị tương ứng là dãy dữ liệu của cột đó. Dùng `print(df)` để xem nội dung bảng.

````python
import pandas as pd

df = pd.DataFrame(
    {
        "numbers": [1, 2, 3],
        "colors": ["red", "white", "blue"],
    }
)
print(df)
````

Dùng tham số `columns` để chỉ định thứ tự các cột.

````python
df = pd.DataFrame(
    {"numbers": [1, 2, 3], "colors": ["red", "white", "blue"]},
    columns=["numbers", "colors"],
)
print(df)
````

### Bài tập 1. Phân tích khám phá bộ dữ liệu IRIS

1. Đọc bộ dữ liệu Iris từ tệp `iris.csv`.
2. Hiển thị 5 điểm dữ liệu đầu tiên.
3. Tự tính số lượng dữ liệu (`count`), trung bình (`mean`), độ lệch chuẩn (`std`), giá trị nhỏ nhất (`min`) và lớn nhất (`max`) cho bốn cột số. Tạo một DataFrame trình bày kết quả theo mẫu sau.

**Ràng buộc:** Không dùng các hàm tính thống kê có sẵn để tính các đại lượng trên. Tự thực hiện phép đếm, cộng dồn, so sánh và tính độ lệch chuẩn; không dùng `count`, `sum`, `mean`, `std`, `min`, `max`, `describe` hoặc các hàm tương đương của NumPy/pandas cho phần tính toán.

| | sepal_length | sepal_width | petal_length | petal_width |
|---|---|---|---|---|
| count | ? | ? | ? | ? |
| mean | ? | ? | ? | ? |
| std | ? | ? | ? | ? |
| min | ? | ? | ? | ? |
| max | ? | ? | ? | ? |

Cột `species` là nhãn loài, không nằm trong bảng thống kê số cần tính.

````python
# TODO - Hoàn thành Bài tập 1
# Đọc iris.csv và hiển thị 5 dòng đầu.
data = pd.read_csv("iris.csv")
print("5 Dong dau")
print(data.head(5),"\n")


# Tự tính count, mean, std, min, max cho bốn cột số.
def count_func(data):
    count = 0
    for x in data:
        count += 1
    return count


def mean_func(data):
    sum = 0
    for x in data:
        sum += x
    return sum / count_func(data)


def std_func(data):
    # tong binh phuoong do lech so voi trung binh
    sum_sq_diff = 0
    for x in data:
        sum_sq_diff += (x - mean_func(data)) ** 2
    # phuong sai
    variance = sum_sq_diff / (count_func(data) - 1)
    # do lech chuan
    std = math.sqrt(variance)
    return std


def min_func(data):
    min_val = data.iloc[0]
    for i in data:
        if i < min_val:
            min_val = i
    return min_val


def max_func(data):
    max_val = data.iloc[0]
    for i in data:
        if i > max_val:
            max_val = i
    return max_val


# Tạo DataFrame kết quả theo mẫu đề bài.
cols = ["sepal_length", "sepal_width", "petal_length", "petal_width"]
results = {}
for col in cols:
    cols_data = data[col]
    count = count_func(cols_data)
    mean = mean_func(cols_data)
    std = std_func(cols_data)
    min = min_func(cols_data)
    max = max_func(cols_data)
    results[col] = [count, mean, std, min, max]

df = pd.DataFrame(results, index=["count", "mean", "std","min","max"])
print(df.round(2))
````
