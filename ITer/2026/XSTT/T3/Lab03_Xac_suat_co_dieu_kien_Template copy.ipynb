## Phần I. Lý Thuyết

### 1. Xác suất có điều kiện

Xác suất của một biến cố có thể phụ thuộc vào nhiều yếu tố và điều kiện. Chẳng hạn, khi chọn ngẫu nhiên hai sinh viên tham gia một trò chơi, xác suất sinh viên thứ hai là nữ có thể phụ thuộc vào kết quả của lần chọn thứ nhất.

Khái niệm **xác suất có điều kiện** mô tả xác suất của biến cố $A$ khi biết biến cố $B$ đã xảy ra. Trong một không gian xác suất, nếu $P(B)>0$ thì:

$$P(A\mid B)=\frac{P(A\cap B)}{P(B)}.$$


Hệ quả trực tiếp là:

$$P(A\cap B)=P(A\mid B)\,P(B).$$

### 1.1. Bài toán 1

Ông Smith có hai người con và biết rằng có ít nhất một người con trai. Tính xác suất cả hai người con đều là con trai.

Gọi $\Omega=\{\text{BB},\text{BG},\text{GB},\text{GG}\}$ là không gian mẫu về giới tính của hai người con, trong đó `B` là con trai và `G` là con gái.

- $A=\{\text{BB}\}$: cả hai người con đều là con trai.
- $C=\{\text{BB},\text{BG},\text{GB}\}$: có ít nhất một người con trai.

Nếu chưa biết điều kiện, $P(A)=1/4$. Khi biết $C$ đã xảy ra, trường hợp `GG` bị loại và không gian đang xét chỉ còn ba kết quả `BB`, `BG`, `GB`. Chỉ `BB` thỏa $A$, nên:

$$P(A\mid C)=\frac{1}{3}.$$

Có thể kiểm tra bằng công thức:

$$P(A\mid C)=\frac{P(A\cap C)}{P(C)}=\frac{1/4}{3/4}=\frac{1}{3}.$$

Xác suất thay đổi từ $1/4$ thành $1/3$ không phải vì khả năng sinh con trai thay đổi, mà vì thông tin mới đã loại bỏ kết quả không phù hợp.

````python
from fractions import Fraction

def P(event, space):
    """Xác suất của một biến cố trong không gian mẫu đồng khả năng."""
    return Fraction(len(event & space), len(space))

S_children = {'BB', 'BG', 'GB', 'GG'}
C_children = {s for s in S_children if 'B' in s}#It nhat 1 nguoi con trai (B)
A_intersect_C_children = {s for s in C_children if s.count('B') == 2}

P_C_children = P(C_children, S_children)
P_A_intersect_C_children = P(A_intersect_C_children, S_children)
P_A_given_C_children = P_A_intersect_C_children / P_C_children

print('Không gian sau khi biết C:', sorted(C_children))
print('A ∩ C:', sorted(A_intersect_C_children))
print('P(A|C) =', P_A_given_C_children)
````

### 1.2. Bài toán 2

Một lớp có 17 sinh viên, gồm 10 nữ và 7 nam. Có 3 sinh viên tên Thanh, gồm 1 nữ và 2 nam. Giáo viên gọi ngẫu nhiên một sinh viên lên bảng. Biết người được chọn là nữ, xác suất người đó tên Thanh là bao nhiêu?



Gọi $A$ là biến cố "sinh viên được gọi tên Thanh" và $B$ là biến cố "sinh viên được gọi là nữ". Không gian mẫu $\Omega$ có 17 phần tử đồng khả năng; $A$ có 3 phần tử, $B$ có 10 phần tử và $A\cap B$ có 1 phần tử. Do đó:

$$P(A)=\frac{\#A}{\#\Omega}=\frac{3}{17},\qquad P(B)=\frac{\#B}{\#\Omega}=\frac{10}{17},$$

$$P(A\cap B)=\frac{\#(A\cap B)}{\#\Omega}=\frac{1}{17},$$

$$P(A\mid B)=\frac{P(A\cap B)}{P(B)}=\frac{1/17}{10/17}=\frac{1}{10}.$$

Ở đây, thông tin "sinh viên được gọi là nữ" thu hẹp không gian từ 17 sinh viên xuống 10 sinh viên nữ; trong nhóm đó chỉ có 1 người tên Thanh.

Thứ tự của điều kiện rất quan trọng. $P(\text{Thanh}\mid\text{Nữ})=1/10$ hỏi "biết là nữ, xác suất tên Thanh là bao nhiêu?" Trong khi $P(\text{Nữ}\mid\text{Thanh})=1/3$ hỏi "biết tên Thanh, xác suất là nữ là bao nhiêu?". Vì vậy:

$$P(A\mid B)\ne P(B\mid A).$$

### 1.3. Bộ phân loại dựa trên xác suất có điều kiện

Xác suất có điều kiện có thể được dùng trực tiếp để tạo một quy tắc phân loại đơn giản. Xét dữ liệu thời tiết và quyết định chơi bóng:

| Weather | Play football? |
|---|---|
| Sunny | Yes |
| Sunny | Yes |
| Sunny | No |
| Rainy | No |
| Rainy | No |
| Rainy | Yes |
| Cloudy | Yes |
| Cloudy | Yes |

Nếu hôm nay trời `Sunny`, ta so sánh hai xác suất có điều kiện. Có 3 ngày nắng, trong đó 2 ngày chơi bóng và 1 ngày không chơi:

$$P(\text{Yes}\mid\text{Sunny})=\frac{2}{3},\qquad P(\text{No}\mid\text{Sunny})=\frac{1}{3}.$$

Vì $2/3>1/3$, quy tắc dự đoán là:

$$\boxed{\text{Sunny}\rightarrow\text{Yes}}$$

Đây là một **bộ phân loại theo quy tắc được ước lượng từ tần suất có điều kiện**. Nó minh họa cách dùng xác suất có điều kiện trong thuật toán, nhưng chưa phải Naive Bayes.

````python
data = [
    ('Sunny', 'Yes'),
    ('Sunny', 'Yes'),
    ('Sunny', 'No'),
    ('Rainy', 'No'),
    ('Rainy', 'No'),
    ('Rainy', 'Yes'),
    ('Cloudy', 'Yes'),
    ('Cloudy', 'Yes'),
]

sunny = [label for weather, label in data if weather == 'Sunny']
p_yes = sunny.count('Yes') / len(sunny)
p_no = sunny.count('No') / len(sunny)
prediction = 'Yes' if p_yes > p_no else 'No'

print(f'P(Yes | Sunny) = {p_yes:.4f}')
print(f'P(No  | Sunny) = {p_no:.4f}')
print('Prediction:', prediction)
````

Kết quả:

```text
P(Yes | Sunny) = 0.6667
P(No  | Sunny) = 0.3333
Prediction: Yes
```

Trong dữ liệu đồng khả năng, phép tính trên chính là:

$$P(\text{Play}\mid\text{Sunny})=\frac{\#(\text{Play}\cap\text{Sunny})}{\#\text{Sunny}}=\frac{2}{3}.$$
