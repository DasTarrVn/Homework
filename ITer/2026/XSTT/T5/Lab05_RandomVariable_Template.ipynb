## Phần I. Lý Thuyết

### 1. Định nghĩa

“Biến” là một đại lượng có thể thay đổi, còn “ngẫu nhiên” nghĩa là chưa biết trước. Vì vậy, **biến ngẫu nhiên** là một đại lượng có thể thay đổi và không thể xác định trước giá trị trước khi thực hiện phép thử.

Trong phép thử tung xúc xắc, ta định nghĩa biến ngẫu nhiên $X = \{1, 2, 3, 4, 5, 6\}$ tương ứng với kết quả xuất hiện mặt 1, 2, 3, 4, 5 hoặc 6. Kết quả của mỗi lần tung là ngẫu nhiên.

Ví dụ sau mô phỏng 8.000 lần tung xúc xắc và xác định tập giá trị quan sát được của $X$.

````python
import random

# random.seed(42)
x = [random.randint(1, 6) for _ in range(8000)]
X = set(x)
print(x[:20])
print(X)
````

### 2. Phân loại biến ngẫu nhiên

Biến ngẫu nhiên được chia thành hai loại:

- **Biến ngẫu nhiên rời rạc:** có tập giá trị hữu hạn hoặc vô hạn đếm được $x_1, x_2, \ldots$. Nói cách khác, ta có thể liệt kê các giá trị của biến ngẫu nhiên.
- **Biến ngẫu nhiên liên tục:** có thể nhận mọi giá trị trong một khoảng giá trị nhất định.

### 3. Phân phối xác suất

Phân phối xác suất của một biến ngẫu nhiên là quy tắc gán một xác suất cho mỗi giá trị có thể có của biến ngẫu nhiên đó.

Với dữ liệu mô phỏng phép thử tung xúc xắc ở trên, xác suất thực nghiệm của mỗi giá trị được tính bằng tần số xuất hiện chia cho tổng số lần thử.

````python
X = sorted(set(x))
print(X)
P = [x.count(value) / len(x) for value in X]
print(P)
````

### 4. Hàm phân phối tích lũy - CDF

Hàm phân phối tích lũy mô tả đầy đủ phân phối xác suất của biến ngẫu nhiên thực $X$. Với mỗi số thực $x$, CDF được định nghĩa bởi

$$
F(x) = P(X \le x) = \sum_{x_i \le x} p_i, \qquad x \in \mathbb{R}.
$$

Nếu gọi các mặt 1, 2, 3 là “nhỏ” và các mặt 4, 5, 6 là “lớn”, xác suất nhận kết quả nhỏ là $P(X \le 3)$.

````python
FX = sum(P[:3])
print(FX)
````

### 5. Các tham số đặc trưng của biến ngẫu nhiên

#### 5.1. Kỳ vọng

Giả sử biến ngẫu nhiên $X$ nhận một trong các giá trị $x_1, x_2, \ldots, x_n$ với các xác suất tương ứng $p_1, p_2, \ldots, p_n$. Kỳ vọng của $X$, ký hiệu $\mu$ hoặc $E(X)$, là

$$
\mu = E(X) = \sum_{i=1}^{n} x_i p_i.
$$

````python
EX = 0
for value, probability in zip(X, P):
    EX += value * probability
print(EX)
````

#### 5.2. Phương sai

Phương sai đo mức độ phân tán của các giá trị của biến ngẫu nhiên quanh kỳ vọng:

$$
\operatorname{Var}(X) = \sum_{i=1}^{n}(x_i - \mu)^2 p_i.
$$

````python
VarX = 0
for value, probability in zip(X, P):
    VarX += (value - EX) ** 2 * probability
print(VarX)
````

#### 5.3. Độ lệch chuẩn

Phương sai biểu diễn độ phân tán nhưng có đơn vị là bình phương đơn vị của biến ngẫu nhiên. Độ lệch chuẩn có cùng ý nghĩa về độ phân tán và cùng đơn vị với biến ngẫu nhiên:

$$
\sigma = \sqrt{\operatorname{Var}(X)}.
$$

````python
import math

SD = math.sqrt(VarX)
print(SD)
````

#### 5.4. Điểm chuẩn hóa - Standard Score

Độ lệch chuẩn cho biết mức độ phân tán trung bình của toàn bộ tập dữ liệu nhưng không cho biết vị trí tương đối của một điểm cụ thể. Điểm chuẩn hóa của giá trị $x$ được tính bởi

$$
z = \frac{x - \mu}{\sigma}.
$$

````python
import random


def simulate_die_pmf(n=10_000, seed=42):
    random.seed(seed)
    counts = [0,0,0,0,0,0]  

    for _ in range(n):
        dice = random.randint(1, 6)
        counts[dice - 1] += 1  

    pmf = [c / n for c in counts]

    return counts, pmf
print(simulate_die_pmf(10000,42))
````
