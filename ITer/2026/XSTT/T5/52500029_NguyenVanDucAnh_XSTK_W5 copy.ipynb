### Bài tập 1. Mô phỏng biến ngẫu nhiên rời rạc

Mô phỏng một biến ngẫu nhiên rời rạc biểu diễn kết quả tung một xúc xắc sáu mặt và tính hàm khối xác suất (PMF).

1. Mô phỏng 10.000 lần tung xúc xắc.
2. Đếm số lần xuất hiện của từng kết quả từ 1 đến 6.
3. Tính PMF của các kết quả.

**Đầu ra cần có:** dữ liệu 10.000 lần tung, số lần xuất hiện và xác suất thực nghiệm của từng mặt 1-6.

**Gợi ý:** Sinh các số nguyên trong đoạn từ 1 đến 6; đếm tần số của từng mặt; sau đó chia mỗi tần số cho tổng số lần tung. Có thể dùng `numpy` hoặc mô-đun `random`.

````python
import random


def simulate_die_pmf(n=10_000, seed=42):
    # Trả về dữ liệu mô phỏng, các mặt, tần số và PMF.
    du_lieu_mo_phong = []

    tan_so = [0, 0, 0, 0, 0, 0]
    for _ in range(n):
        dice = random.randint(1, 6)
        du_lieu_mo_phong.append(dice)
        tan_so[dice - 1] += 1

    mat = set(du_lieu_mo_phong)
    pmf = [s / n for s in tan_so]
    return du_lieu_mo_phong[:20], mat, tan_so, pmf


print(simulate_die_pmf(10_000, 42))
````

### Bài tập 2. Biến ngẫu nhiên liên tục và PDF

Định nghĩa một biến ngẫu nhiên liên tục có phân phối chuẩn với trung bình 0 và độ lệch chuẩn 1. Tính hàm mật độ xác suất (PDF).

1. Dùng `scipy.stats.norm` để định nghĩa phân phối(Distribution).
2. Tạo các giá trị $x$ từ -4 đến 4.
3. Tính PDF tại các giá trị $x$ đó.

**Đầu ra cần có:** mảng giá trị $x$ và mảng giá trị PDF tương ứng.

**Gợi ý:** Tạo đối tượng phân phối bằng `norm` với hai tham số vị trí và tỷ lệ; tạo một lưới số đều bằng `numpy.linspace`; sau đó gọi phương thức tính mật độ của đối tượng phân phối.

````python
# TODO - Hoàn thành Bài tập 2
import numpy as np
from scipy.stats import norm


def standard_normal_pdf(num_points=100):
    distribution = norm(loc=0, scale=1)
    x = np.linspace(-4, 4, num_points)
    pdf= distribution.pdf(x)
    return x,pdf
x, pdf = standard_normal_pdf(num_points=100)
for  a, b in zip(x,pdf):
    print(f"{a:4f} : {b:4f}")
````

### Bài tập 3. Hàm phân phối tích lũy - CDF

Tính hàm phân phối tích lũy của phân phối đều trên đoạn $[0, 1]$.

1. Dùng `numpy` để tạo một mảng các giá trị $x$ từ 0 đến 1.
2. Tính CDF theo công thức

$$
F(x) = \frac{x-a}{b-a}.
$$

Với bài này, $a=0$ và $b=1$.

**Đầu ra cần có:** mảng các giá trị $x$ và mảng CDF tương ứng.

**Gợi ý:** Tạo một lưới số đều trong $[a,b]$, rồi áp dụng trực tiếp công thức cho toàn bộ mảng nhờ phép toán vector của `numpy`.

````python
# TODO - Hoàn thành Bài tập 3
import numpy as np


def uniform_cdf(a=0.0, b=1.0, num_points=101):
    x = np.linspace(a, b, num_points)
    cdf= (x-a)/(b-a)
    return x,cdf
x,cdf =uniform_cdf()
print(x)
print(cdf)
````

### Bài tập 4. Kỳ vọng của biến ngẫu nhiên rời rạc

Tính kỳ vọng của biến ngẫu nhiên rời rạc được xác định bởi PMF sau:

- $P(X=1)=0.2$
- $P(X=2)=0.5$
- $P(X=3)=0.3$

1. Khai báo các giá trị của biến ngẫu nhiên và các xác suất tương ứng.
2. Tính kỳ vọng $E(X)$.
3. In kết quả.

**Đầu ra cần có:** giá trị kỳ vọng $E(X)$.

**Gợi ý:** Nhân từng giá trị có thể có của $X$ với xác suất tương ứng rồi cộng tất cả các tích, theo công thức $E(X)=\sum_i x_i p_i$.

````python
# Hoàn thành Bài tập 4
import numpy as np


def discrete_expectation():
    x_vals = np.array([1, 2, 3], dtype=float)
    p_vals = np.array([0.2, 0.5, 0.3])

    exp_val = np.sum(x_vals * p_vals)
    return exp_val


expectation = discrete_expectation()
print(expectation)
````

### Bài tập 5. Phương sai của biến ngẫu nhiên

Tính phương sai của biến ngẫu nhiên đã cho trong Bài tập 4.

1. Dùng cùng PMF của Bài tập 4.
2. Tính phương sai theo công thức

$$
\operatorname{Var}(X)=E(X^2)-[E(X)]^2.
$$

3. In phương sai.

**Đầu ra cần có:** giá trị phương sai của $X$.

**Gợi ý:** Tính $E(X)$ và $E(X^2)$ riêng. Với $E(X^2)$, bình phương từng giá trị của $X$ trước khi nhân với xác suất tương ứng.

````python
# Hoàn thành Bài tập 5
import numpy as np


def discrete_variance():
    x_vals = np.array([1, 2, 3], dtype=float)
    p_vals = np.array([0.2, 0.5, 0.3])

    exp_x = np.sum(x_vals * p_vals)
    exp_x2 = np.sum((x_vals**2) * p_vals)

    var_x = exp_x2 - (exp_x**2)
    return var_x


variance = discrete_variance()
print(variance)
````

### Bài tập 6. Số ca cấp cứu trong ngày

Số ca cấp cứu tại một bệnh viện trong một ngày là một biến ngẫu nhiên. Bệnh viện thu thập dữ liệu trong 10 năm và lưu vào biến `x` như sau:

```python
x = np.random.choice(
    [0, 1, 2, 3, 4, 5],
    3650,
    p=[0.1, 0.2, 0.3, 0.2, 0.15, 0.05],
)
```

Mỗi giá trị trong `x` là số ca cấp cứu trong một ngày. Ví dụ, `x = [0, 5, 2]` biểu diễn dữ liệu của ba ngày với số ca tương ứng là 0, 5 và 2.

Thực hiện các yêu cầu:

(a) Xác định các giá trị của biến ngẫu nhiên $X$ và lưu vào biến `X`.

(b) Tính phân phối xác suất thực nghiệm của $X$ và lưu vào biến `P` kiểu danh sách.

(c) Tính kỳ vọng, phương sai và độ lệch chuẩn của $X$.

(d) Tính xác suất có từ 3 ca cấp cứu trở lên.

**Đầu ra cần có:** `X`, `P`, kỳ vọng, phương sai, độ lệch chuẩn và xác suất ở câu (d).

**Gợi ý:** Lấy các giá trị không trùng nhau trong `x` và sắp xếp; với mỗi giá trị, đếm số ngày xuất hiện rồi chia cho 3.650. Dùng `X` và `P` cùng các công thức ở phần lý thuyết. Ở câu (d), cộng các xác suất tương ứng với những giá trị không nhỏ hơn 3.

````python
# Hoàn thành Bài tập 6
import numpy as np

np.random.seed(42)
x = np.random.choice(
    [0, 1, 2, 3, 4, 5],
    3650,
    p=[0.1, 0.2, 0.3, 0.2, 0.15, 0.05],
)

X = sorted(list(set(x)))

n_samples = len(x)
P = [int(np.count_nonzero(x == val)) / n_samples for val in X]

expectation = sum(val * prob for val, prob in zip(X, P))
variance = sum(((val - expectation) ** 2) * prob for val, prob in zip(X, P))
standard_deviation = np.sqrt(variance)

probability_at_least_3 = sum(prob for val, prob in zip(X, P) if val >= 3)

print(X)
print(P)
print(expectation)
print(variance)
print(standard_deviation)
print(probability_at_least_3)
````

### Bài tập 7. Thí nghiệm tung hai đồng xu

Viết chương trình thực hiện 10.000 lần thí nghiệm tung hai đồng xu. Gọi $X$ là số mặt ngửa xuất hiện trong mỗi lần thí nghiệm.

(a) Lưu kết quả vào biến `x` kiểu danh sách. Ví dụ, `x = [0, 2, 1]` biểu diễn ba lần thí nghiệm: lần đầu không có mặt ngửa, lần thứ hai cả hai đồng xu đều ngửa, lần thứ ba có đúng một mặt ngửa.

(b) Tìm các giá trị của biến ngẫu nhiên $X$ và lưu vào biến `X`.

(c) Tính phân phối xác suất thực nghiệm của $X$ và lưu vào biến `P` kiểu danh sách.

(d) Tính kỳ vọng, phương sai và độ lệch chuẩn của $X$.

(e) Tính xác suất có ít nhất một mặt ngửa.

**Đầu ra cần có:** `x`, `X`, `P`, kỳ vọng, phương sai, độ lệch chuẩn và xác suất ở câu (e).

**Gợi ý:** Mỗi đồng xu có thể được mã hóa bằng 0/1; tổng của hai kết quả là số mặt ngửa. Có thể mô phỏng trực tiếp số mặt ngửa bằng phân phối nhị thức với hai phép thử. Sau đó xử lý `X`, `P` và các tham số giống Bài tập 6; ở câu (e), cộng xác suất của các giá trị lớn hơn hoặc bằng 1.

````python
# Hoàn thành Bài tập 7
import numpy as np


def simulate_two_coins(n=10_000, seed=42):
    rng = np.random.default_rng(seed)
    flips = rng.binomial(n=2, p=0.5, size=n).tolist()

    distinct_vals = sorted(list(set(flips)))

    total_experiments = len(flips)
    emp_p = [flips.count(k) / total_experiments for k in distinct_vals]

    exp_val = sum(k * p for k, p in zip(distinct_vals, emp_p))
    var_val = sum(((k - exp_val) ** 2) * p for k, p in zip(distinct_vals, emp_p))
    std_val = np.sqrt(var_val)

    p_at_least_1 = sum(p for k, p in zip(distinct_vals, emp_p) if k >= 1)

    return (
        flips,
        distinct_vals,
        emp_p,
        exp_val,
        var_val,
        std_val,
        p_at_least_1,
    )


(
    x,
    X,
    P,
    expectation,
    variance,
    standard_deviation,
    probability_at_least_one_head,
) = simulate_two_coins()

print(x[:20])
print(X)
print(P)
print(expectation)
print(variance)
print(standard_deviation)
print(probability_at_least_one_head)
````
